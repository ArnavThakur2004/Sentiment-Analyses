# removing Stop words (a, the , of , are)

```python
import nltk 
from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
    new_text = []

    for word in text.split():
        if word in stopwords.words("english"):
            new_text.append("")
        else:
            new_text.append(word)
    
    x = new_text[:]
    new_text.clear()
    return " ".join(x)

df["text"].apply(remove_stopwords)
```
